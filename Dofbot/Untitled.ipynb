```python
import cv2
import ipywidgets as widgets
from threading import Thread, Event
import time

# Function to convert BGR image to JPEG format
def bgr8_to_jpeg(value, quality=75):
    return bytes(cv2.imencode('.jpg', value)[1])

# Shared event to signal image capture
capture_event = Event()

# Function to handle the camera feed
def camera_feed_handler(image, image_widget):
    while True:
        ret, frame = image.read()
        if ret:
            image_widget.value = bgr8_to_jpeg(frame)

        if capture_event.is_set():
            # Capture and save image when the event is set
            cv2.imwrite('captured_image.jpg', frame)
            print("Picture captured!")
            capture_event.clear()

# Set up camera and image widget
image_widget = widgets.Image(format='jpeg', width=600, height=500)
display(image_widget)

image = cv2.VideoCapture(0)

# Start the camera feed in a separate thread
thread = Thread(target=camera_feed_handler, args=(image, image_widget))
thread.start()

# Main program loop
try:
    while True:
        command = input("Enter 'capture' to take a picture, 'exit' to quit: ").strip().lower()
        if command == 'capture':
            capture_event.set()
        elif command == 'exit':
            break
except KeyboardInterrupt:
    print("Program closed!")

# Clean up
image.release()
thread.join()
```
