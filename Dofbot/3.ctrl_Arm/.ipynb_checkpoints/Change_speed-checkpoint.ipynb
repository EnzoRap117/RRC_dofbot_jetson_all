```python

def main():
    Arm.Arm_serial_servo_write6(90, 90, 90, 90, 90, 90, 500)
    time.sleep(1)
    
    while(True):
        print("please select: \n1 -> change speed manually\
                \n2 -> change speed automatically\
                \n3 -> exit the program")
        mode = input()
        if mode == "1":
            manual_speed()
        elif mode == "2": 
            auto_speed()
        elif mode == "3": 
            break
        else:
            print("Unknown input")

# convert speed from degree/second to millisecond
def speed_conversion(speed, start_angle, desired_angle):
    step = abs(desired_angle - start_angle)
    time = step/speed
    milli_time = round(time * 1000)
    return milli_time

def move_servo(desired_angle, servo_number, speed):
    # read the current angle of servo 5
    
    print("enter move servo")
    print(servo_number)
    
    current_angle = Arm.Arm_serial_servo_read(servo_number)
    
    print("read servo succ")
    
    Arm.Arm_serial_servo_write(servo_number, desired_angle, speed_conversion(speed, current_angle, desired_angle))
    # Wait until the servo reaches the desired angle
    while(Arm.Arm_serial_servo_read(servo_number) != desired_angle):
        pass
    print("The servo %d reached the desired angle" % (servo_number))

def manual_speed():
    while(True):
        print("please input the speed in degree/second(range: 1-45)")
        speed = int(input())
        
        print(speed)
        
        if(1 <= speed <= 45):
            
            print("enter if")
            
            desired_angle = 45
            servo_number = 5
            move_servo(desired_angle, servo_number, speed)
            print("Press \"r\" to reset the servo to 90 degree,\nor press any key else to select another mode")
            end_choice = input()
            if end_choice == "r":
                print("enter if 2")
                Arm.Arm_serial_servo_write(servo_number, 90, 500)
                print("write servo")
            else: 
                print("enter pass")
                pass
        else:
            print("speed out of range: 1-45")
def auto_speed():
    while(True): 
        desired_angle = 45
        servo_number = 5
        print("Press \"s\" to start, press any key else to select another mode")
        start = input()
        if start == "s":
            while(True):
                speed = random.randint(1, 45)
                print("random speed: %d" % (speed))
                move_servo(desired_angle, servo_number, speed)
                print("The servo reached desired angle %d with speed %d" % (desired_angle, speed))
                print("Press \"e\" to stop, or press any key else to generate another random speed")
                end_choice = input()
                if end_choice == "e": 
                    break
        else:
            break
try :
    main()
except KeyboardInterrupt:
    print(" Program closed! ")
    pass
```
