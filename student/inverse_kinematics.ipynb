```python
def rotz(q):
    q_rad = q*math.pi/180
    cos = math.cos(q_rad)
    sin = math.sin(q_rad)
    return np.array([[cos, -1*sin, 0] , [sin, cos, 0] , [0, 0, 1]])

def roty(q):
    q_rad = q*math.pi/180
    cos = math.cos(q_rad)
    sin = math.sin(q_rad)
    return np.array([[cos, 0, sin], [0, 1, 0], [-1*sin, 0, cos]])

def rotx(q):
    q_rad = q*math.pi/180
    cos = math.cos(q_rad)
    sin = math.sin(q_rad)
    return np.array([[1, 0, 0], [0, cos, -1*sin], [0, sin, cos]])

def rot(k, theta):
    """
    Generates a 3 x 3 rotation matrix from a unit 3 x 1 unit vector axis
    and an angle in radians using the Euler-Rodrigues formula
    
        R = I + sin(theta)*hat(k) + (1 - cos(theta))*hat(k)^2
        
    :type    k: numpy.array
    :param   k: 3 x 1 unit vector axis
    :type    theta: number
    :param   theta: rotation about k in radians
    :rtype:  numpy.array
    :return: the 3 x 3 rotation matrix 
        
    """
    I = np.identity(3)
    khat = hat(k)
    khat2 = khat.dot(khat)
    return I + math.sin(theta)*khat + (1.0 - math.cos(theta))*khat2
```

```python
def fwdkin(q):
    P01 = ez * (l0+l1)
    P12 = np.zeros([3])
    P23 = ex * l2
    P34 = ez * -l3
    P45 = np.zeros([3])
    P5T = ex * -(l4+l5)
    
    R01 = rotz(q.item(0))
    R12 = roty(-q.item(1))
    R23 = roty(-q.item(2))
    R34 = roty(-q.item(3))
    R45 = rotx(-q.item(4))
    R5T = np.identity(3)

    Rot = R01 @ R12 @ R23 @ R34 @ R45 @ R5T

    Pot = P01 + R01 @ ( P12 + R12 @ ( P23 + R23 @ ( P34 + R34 @ ( P45 + R45 @ P5T))))

    return [Rot, Pot]
```

```python
def pathplanPID(qf, speedtime):
    # q0 - initial configuration
    q0 = getConfig()

    # qf - final configuration (target)
    f = open("pathplanPID.txt", "w+")
    # Create joint space path
    points = 20
    lmbda = np.arange(0, 1+(1/points), 1/points)
    Kp = 0.5
    Ki = 0
    Kd = 0.2
    qprev = 0
    qcumulative = 0
    prevtime = time.time()
    qact=q0
    for lm in lmbda:
        qdes = (1-lm) * q0 + lm * qf
        qerror = qdes - qact
        qcumulative = qcumulative + qerror
        curtime = time.time()
        Ts = curtime - prevtime
        Up = qerror * Kp
        Ui = Ki * Ts * qcumulative
        Ud = np.array([0,0,0,0,0,0])
        if lm != 0:
            Ud = Kd * (qerror - qprev) / Ts
        qeff = qdes + Up + Ui + Ud
        for q in qeff:
            if q >180:
                q=180
            elif q<0:
                q=0
#         print("qdes: ", qdes)
#         print("qeff: ", qeff)
#         print("Up: ", Up)
#         print("Ui: ", Ui)
#         print("Ud: ", Ud)
        prevtime = curtime
        Arm.Arm_serial_servo_write6(qeff[0],qeff[1],qeff[2],qeff[3],qeff[4],qeff[5], speedtime)
        time.sleep(.1)
        qact = getConfig()
        f.write("q1,{:.2f},{},".format(qdes[0], qact[0]))
        f.write("q2,{:.2f},{},".format(qdes[1], qact[1]))
        f.write("q3,{:.2f},{},".format(qdes[2], qact[2]))
        f.write("q4,{:.2f},{},".format(qdes[3], qact[3]))
        f.write("q5,{:.2f},{},".format(qdes[4], qact[4]))
        f.write("q6,{:.2f},{}\n".format(qdes[5], qact[5]))
        qprev = qerror
```

```python
def invkin(Rot, Pot):
#     q = np.matmul(np.nan, np.ones((5,4)))
    q = np.full((5,4),np.nan)

    k = -ey
    h = ez
    p = ex
    d = ez.T @ Rot @ ex
    thetatmp = sp4(h.T, p, k, d)
    
    if len(thetatmp) == 1:
        theta = [thetatmp[0], np.nan, thetatmp[0], np.nan]
    else:
        theta = [thetatmp[0], thetatmp[1], thetatmp[0], thetatmp[1]]
        
#     print("Theta: ", theta)

    # Get q1 from Rot
    for ii in range(4):
        if not np.isnan(theta[ii]):
            k = ez
            p1 = np.matmul(rot(-ey,theta[ii]), ex)
            p2 = np.matmul(Rot, ex)
            q[0][ii] = sp1(p1,p2,k)
            
    # Get q5 from Rot
    for ii in range(4):
        if not np.isnan(theta[ii]):
            k = ex
            p1 = rot(ey, theta[ii]) @ ez
            p2 = np.transpose(Rot) @ ez
            q[4][ii] = sp1(p1,p2,k)
            
    # Get q3 from Pot
    for ii in range(2):
        if not np.isnan(theta[ii]):
            temp1 = np.matmul(rot(ez, -q[0][ii]),(Pot - L1*ez))
            temp2 = np.matmul(rot(ey, -theta[ii]),L4*ex)
            Pprime =  temp1 + temp2
            k = -ey
            p1 = l3 * ez
            p2 = l2 * ex
            d = np.linalg.norm(Pprime)
            q3tmp = sp3(p1,p2,k,d)
            if len(q3tmp) == 1:
                q[2][ii] = q3tmp[0]
            elif len(q3tmp) == 2:
                q[2][ii] = q3tmp[0]
                q[2][ii+2] = q3tmp[1]
            else:
                print("fuck off")
                
                
    # get q2 from Pot
    for ii in range(4):
        if not np.isnan(q[2][ii]):
            temp1 = np.matmul(rot(ez, -q[0][ii]) , (Pot-L1*ez))
            temp2 = np.matmul(rot(ey, -theta[ii]), L4*ex)
            Pprime = ( temp1 )+( temp2 )
            k = -ey
            temp = np.matmul(rot(-ey, q[2][ii]), l3*ez)
            p1 = l2*ex - temp
            p2 = Pprime
            q[1][ii] = sp1(p1,p2,k)

    # get q4 from theta
    for ii in range(4):
        if not np.isnan(q[2][ii]):
            q[3][ii] = theta[ii]-q[1][ii]-q[2][ii]
            
    # remove nonvalid columns of q
    for ii in range(3,-1,-1):
        sum = 0
        for jj in q:
            sum = sum + (np.isnan(jj[ii]))
        if sum:
            q = np.delete(q, ii, 1)
            
    q = q*180/np.pi
    
    qout_of_range = np.zeros(len(q[1,:]))
    for ii in range(0,len(q)):
        for jj in range(0,len(q[ii])):
            while q[ii][jj] < -180:
                q[ii][jj] = q[ii][jj] + 360
            while q[ii][jj] > 360:
                q[ii][jj] = q[ii][jj] - 360
            if q[ii][jj]<0:
                qout_of_range[jj] = max(qout_of_range[jj],-q[ii][jj])
            elif q[ii][jj]>180:
                qout_of_range[jj] = max(qout_of_range[jj],q[ii][jj]-180)
    sort_order = qout_of_range.argsort()
    print("sort order: ",sort_order)
    
    print("final q: ",q)
    return q[:, sort_order]
```

```python
def sp0(p, q, k):
    """
    Solves canonical geometric subproblem 0, theta subtended between p and q according to
    
        q = rot(k, theta)*p
           ** assumes k'*p = 0 and k'*q = 0
           
    Requires that p and q are perpendicular to k. Use subproblem 1 if this is not
    guaranteed.

    :type    p: numpy.array
    :param   p: 3 x 1 vector before rotation
    :type    q: numpy.array
    :param   q: 3 x 1 vector after rotation
    :type    k: numpy.array
    :param   k: 3 x 1 rotation axis unit vector
    :rtype:  number
    :return: theta angle as scalar in radians
    """
    
    eps = np.finfo(np.float64).eps
    assert (np.dot(k,p) < eps) and (np.dot(k,q) < eps), \
           "k must be perpendicular to p and q"
    
    norm = np.linalg.norm
    
    ep = p / norm(p)
    eq = q / norm(q)
    
    theta = 2 * np.arctan2( norm(ep - eq), norm (ep + eq))
    
    if (np.dot(k,np.cross(p , q)).any() < 0):
        return -theta
        
    return theta
```

```python
def sp1(p, q, k):
    """
    Solves canonical geometric subproblem 1, theta subtended between p and q according to
    
        q = rot(k, theta)*p
    
    :type    p: numpy.array
    :param   p: 3 x 1 vector before rotation
    :type    q: numpy.array
    :param   q: 3 x 1 vector after rotation
    :type    k: numpy.array
    :param   k: 3 x 1 rotation axis unit vector
    :rtype:  number
    :return: theta angle as scalar in radians
    """
    
    eps = np.finfo(np.float64).eps
    norm = np.linalg.norm
    
    if norm (np.subtract(p, q)) < np.sqrt(eps):
        return 0.0
    
    
    k = np.divide(k,norm(k))
    
    pp = np.subtract(p,np.dot(p, k)*k)
    qp = np.subtract(q,np.dot(q, k)*k)
    
    epp = np.divide(pp, norm(pp))    
    eqp = np.divide(qp, norm(qp))
    
    theta = sp0(epp, eqp, k)
    
    if (np.abs(norm(p) - norm(q)) > norm(p)*1e-2):
        print("||p|| and ||q|| must be the same!!!")
    
    return theta
```

```python
def sp3(p, q, k, d):
    """
    Solves canonical geometric subproblem 3,solve for theta in
    an elbow joint according to
    
        || q + rot(k, theta)*p || = d
        
    may have 0, 1, or 2 solutions
    
    :type    p: numpy.array
    :param   p: 3 x 1 position vector of point p
    :type    q: numpy.array
    :param   q: 3 x 1 position vector of point q
    :type    k: numpy.array
    :param   k: 3 x 1 rotation axis for point p
    :type    d: number
    :param   d: desired distance between p and q after rotation
    :rtype:  list of numbers
    :return: list of valid theta angles in radians        
    
    """
    
    norm=np.linalg.norm
    
    pp = np.subtract(p,np.dot(np.dot(p, k),k))
    qp = np.subtract(q,np.dot(np.dot(q, k),k))
    dpsq = d**2 - ((np.dot(k, np.add(p,q)))**2)
    
    bb=-(np.dot(pp,pp) + np.dot(qp,qp) - dpsq)/(2*norm(pp)*norm(qp))
    
    if dpsq < 0 or np.abs(bb) > 1:
        print("No solution - no rotation can achieve specified distance")
        return []
    
    theta = sp1(pp/norm(pp), qp/norm(qp), k)
    
    phi = np.arccos(bb)
    if np.abs(phi) > 0:
        return [theta + phi, theta - phi]
    else:
        return [theta]
```

```python
def sp4(p, q, k, d):
    """
    Solves canonical geometric subproblem 4, theta for static
    displacement from rotation axis according to
    
        d = p.T*rot(k, theta)*q
        
    may have 0, 1, or 2 solutions
        
    :type    p: numpy.array
    :param   p: 3 x 1 position vector of point p
    :type    q: numpy.array
    :param   q: 3 x 1 position vector of point q
    :type    k: numpy.array
    :param   k: 3x1 rotation axis for point p
    :type    d: number
    :param   d: desired displacement
    :rtype:  list of numbers
    :return: list of valid theta angles in radians    
    """
        
    a = np.matmul(np.matmul(p,hat(k)),q)
    b = -np.matmul(p, np.matmul(hat(k),hat(k).dot(q)))
    c = np.subtract(d, (np.dot(p,q) -b))
    
    phi = np.arctan2(b, a)
    
    d = c / np.linalg.norm([a,b])
    
    if d > 1:
        return []
    
    psi = np.arcsin(d)
    
    return [-phi+psi, -phi-psi+np.pi]
```

```python
def hat(k):
    """
    Returns a 3 x 3 cross product matrix for a 3 x 1 vector
    
             [  0 -k3  k2]
     khat =  [ k3   0 -k1]
             [-k2  k1   0]
    
    :type    k: numpy.array
    :param   k: 3 x 1 vector
    :rtype:  numpy.array
    :return: the 3 x 3 cross product matrix    
    """
    
    khat=np.zeros((3,3))
    khat[0,1]=-k[2]
    khat[0,2]=k[1]
    khat[1,0]=k[2]
    khat[1,2]=-k[0]
    khat[2,0]=-k[1]
    khat[2,1]=k[0]
    
    return khat
```
